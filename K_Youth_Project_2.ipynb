```python
import random
import ipywidgets as widgets
from IPython.display import display, clear_output
import matplotlib.pyplot as plt
from collections import Counter

# ============ ROCK PAPER SCISSORS GAME ============

class RockPaperScissors:
    def __init__(self):
        self.choices = ['👊 Rock', '✋ Paper', '✂️ Scissors']
        self.player_score = 0
        self.computer_score = 0
        self.history = []

        # Create UI elements
        self.output = widgets.Output()

        # Score display
        self.score_label = widgets.Label(
            value=f"Player: {self.player_score}  |  Computer: {self.computer_score}",
            layout=widgets.Layout(width='100%')
        )

        # Buttons
        self.rock_btn = widgets.Button(
            description='👊 Rock',
            button_style='primary',
            layout=widgets.Layout(width='100px', height='80px', margin='5px')
        )

        self.paper_btn = widgets.Button(
            description='✋ Paper',
            button_style='success',
            layout=widgets.Layout(width='100px', height='80px', margin='5px')
        )

        self.scissors_btn = widgets.Button(
            description='✂️ Scissors',
            button_style='warning',
            layout=widgets.Layout(width='100px', height='80px', margin='5px')
        )

        self.reset_btn = widgets.Button(
            description='🔄 Reset Game',
            button_style='danger',
            layout=widgets.Layout(width='150px', height='40px', margin='5px')
        )

        self.result_label = widgets.HTML(
            value="<h3>Choose your move!</h3>",
            layout=widgets.Layout(width='100%')
        )

        # Game stats button
        self.stats_btn = widgets.Button(
            description='📊 Show Statistics',
            button_style='info',
            layout=widgets.Layout(width='150px', height='40px', margin='5px')
        )

        # Button actions
        self.rock_btn.on_click(lambda b: self.play('👊 Rock'))
        self.paper_btn.on_click(lambda b: self.play('✋ Paper'))
        self.scissors_btn.on_click(lambda b: self.play('✂️ Scissors'))
        self.reset_btn.on_click(lambda b: self.reset_game())
        self.stats_btn.on_click(lambda b: self.show_stats())

        # Display UI
        self.display_ui()

    def display_ui(self):
        clear_output(wait=True)

        # Title
        title = widgets.HTML(
            value="<h1>🎮 Rock Paper Scissors Game 🎮</h1>",
            layout=widgets.Layout(width='100%', margin='10px')
        )

        # Instructions
        instructions = widgets.HTML(
            value="<p>Choose your move by clicking one of the buttons below!</p>",
            layout=widgets.Layout(width='100%')
        )

        # Button container
        button_box = widgets.HBox(
            [self.rock_btn, self.paper_btn, self.scissors_btn],
            layout=widgets.Layout(justify_content='center', margin='10px')
        )

        # Control buttons
        control_box = widgets.HBox(
            [self.reset_btn, self.stats_btn],
            layout=widgets.Layout(justify_content='center', margin='5px')
        )

        # Display everything
        display(title, instructions, self.result_label, self.score_label,
                button_box, control_box, self.output)

    def play(self, player_choice):
        computer_choice = random.choice(self.choices)

        # Determine winner
        if player_choice == computer_choice:
            result = "🤝 It's a Tie!"
            points = (0, 0)
        elif (player_choice == '👊 Rock' and computer_choice == '✂️ Scissors') or \
             (player_choice == '✋ Paper' and computer_choice == '👊 Rock') or \
             (player_choice == '✂️ Scissors' and computer_choice == '✋ Paper'):
            result = "🎉 You Win!"
            points = (1, 0)
            self.player_score += 1
        else:
            result = "💻 Computer Wins!"
            points = (0, 1)
            self.computer_score += 1

        # Update history
        self.history.append({
            'player': player_choice,
            'computer': computer_choice,
            'result': result
        })

        # Update display
        with self.output:
            clear_output(wait=True)
            display(widgets.HTML(
                f"""
                <div style='border: 2px solid #ddd; padding: 15px; border-radius: 10px; margin: 10px; background-color: #f9f9f9;'>
                    <h4>Round {len(self.history)}</h4>
                    <p>You chose: <strong>{player_choice}</strong></p>
                    <p>Computer chose: <strong>{computer_choice}</strong></p>
                    <h3 style='color: {'green' if 'Win' in result else 'red' if 'Computer' in result else 'orange'};'>
                        {result}
                    </h3>
                    <p style='font-size: 12px; color: gray;'>Score: You {self.player_score} - {self.computer_score} Computer</p>
                </div>
                """
            ))

        # Update score label
        self.score_label.value = f"Player: {self.player_score}  |  Computer: {self.computer_score}"
        self.result_label.value = f"<h3>Choose your move! (Round {len(self.history)})</h3>"

    def reset_game(self):
        self.player_score = 0
        self.computer_score = 0
        self.history = []
        self.score_label.value = "Player: 0  |  Computer: 0"
        self.result_label.value = "<h3>🔄 Game Reset! Choose your move!</h3>"
        with self.output:
            clear_output()

    def show_stats(self):
        if not self.history:
            with self.output:
                clear_output()
                display(widgets.HTML("<p>No games played yet! Play some rounds first.</p>"))
            return

        # Calculate statistics
        total_games = len(self.history)
        player_wins = sum(1 for h in self.history if 'Win' in h['result'])
        computer_wins = sum(1 for h in self.history if 'Computer' in h['result'])
        ties = total_games - player_wins - computer_wins

        # Player choice distribution
        player_choices = [h['player'] for h in self.history]
        choice_counts = Counter(player_choices)

        # Create statistics display
        with self.output:
            clear_output()

            # Stats text
            stats_html = f"""
            <div style='border: 2px solid #4CAF50; padding: 15px; border-radius: 10px; margin: 10px; background-color: #f0f8ff;'>
                <h4>📊 Game Statistics</h4>
                <p><strong>Total Games:</strong> {total_games}</p>
                <p><strong>Player Wins:</strong> {player_wins} ({player_wins/total_games*100:.1f}%)</p>
                <p><strong>Computer Wins:</strong> {computer_wins} ({computer_wins/total_games*100:.1f}%)</p>
                <p><strong>Ties:</strong> {ties} ({ties/total_games*100:.1f}%)</p>
                <hr>
                <p><strong>Your Choice Distribution:</strong></p>
            """

            for choice, count in choice_counts.items():
                percentage = count/total_games*100
                stats_html += f"<p>{choice}: {count} times ({percentage:.1f}%)</p>"

            stats_html += "</div>"

            # Create a simple matplotlib chart
            fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 4))

            # Pie chart: Win/Loss/Tie
            sizes = [player_wins, computer_wins, ties]
            labels = ['Player Wins', 'Computer Wins', 'Ties']
            colors = ['green', 'red', 'orange']
            ax1.pie(sizes, labels=labels, colors=colors, autopct='%1.1f%%', startangle=90)
            ax1.set_title('Win/Loss/Tie Distribution', fontsize=12)

            # Bar chart: Choice distribution
            choices = list(choice_counts.keys())
            counts = list(choice_counts.values())
            ax2.bar(choices, counts, color=['#FF6B6B', '#4ECDC4', '#45B7D1'])
            ax2.set_title('Your Choice Distribution', fontsize=12)
            ax2.set_ylabel('Count')
            ax2.set_xlabel('Choice')
            plt.tight_layout()
            plt.show()

            display(widgets.HTML(stats_html))

# Start the game
game = RockPaperScissors()
```

```python
import random
import ipywidgets as widgets
from IPython.display import display, clear_output
import time

class NumberGuessingGame:
    def __init__(self):
        self.secret_number = random.randint(1, 100)
        self.attempts = 0
        self.max_attempts = 7
        self.guesses = []
        self.game_over = False

        # UI Elements
        self.output = widgets.Output()

        self.title = widgets.HTML(
            value="<h1>🔢 Number Guessing Game 🎯</h1>",
            layout=widgets.Layout(width='100%')
        )

        self.info = widgets.HTML(
            value="<p>I'm thinking of a number between 1 and 100.<br>You have 7 attempts to guess it!</p>"
        )

        self.guess_input = widgets.IntText(
            description='Your Guess:',
            value=0,
            layout=widgets.Layout(width='200px')
        )

        self.guess_btn = widgets.Button(
            description='Submit Guess',
            button_style='primary',
            layout=widgets.Layout(width='120px', margin='5px')
        )

        self.reset_btn = widgets.Button(
            description='🔄 New Game',
            button_style='danger',
            layout=widgets.Layout(width='120px', margin='5px')
        )

        self.hint_btn = widgets.Button(
            description='💡 Hint',
            button_style='info',
            layout=widgets.Layout(width='100px', margin='5px')
        )

        self.feedback = widgets.HTML(
            value="<p style='color: blue;'>Enter your first guess!</p>"
        )

        self.attempt_label = widgets.Label(
            value=f"Attempts: 0/{self.max_attempts}"
        )

        # Button actions
        self.guess_btn.on_click(self.make_guess)
        self.reset_btn.on_click(self.reset_game)
        self.hint_btn.on_click(self.show_hint)

        # Display UI
        self.display_ui()

    def display_ui(self):
        clear_output(wait=True)

        input_box = widgets.HBox(
            [self.guess_input, self.guess_btn, self.hint_btn, self.reset_btn],
            layout=widgets.Layout(justify_content='center', margin='10px')
        )

        display(self.title, self.info, input_box, self.attempt_label, self.feedback, self.output)

    def make_guess(self, b):
        if self.game_over:
            with self.output:
                clear_output()
                display(widgets.HTML("<p style='color: red;'>Game is over! Click 'New Game' to play again.</p>"))
            return

        guess = self.guess_input.value

        if guess < 1 or guess > 100:
            with self.output:
                clear_output()
                display(widgets.HTML("<p style='color: red;'>⚠️ Please enter a number between 1 and 100!</p>"))
            return

        self.attempts += 1
        self.guesses.append(guess)

        # Check guess
        if guess == self.secret_number:
            self.game_over = True
            feedback_html = f"""
            <div style='border: 2px solid green; padding: 15px; border-radius: 10px; margin: 10px; background-color: #d4edda;'>
                <h3 style='color: green;'>🎉 Correct! You Win!</h3>
                <p>The number was {self.secret_number}.</p>
                <p>You found it in {self.attempts} attempts!</p>
                <p>Your guesses: {self.guesses}</p>
            </div>
            """
            self.feedback.value = feedback_html
        elif self.attempts >= self.max_attempts:
            self.game_over = True
            feedback_html = f"""
            <div style='border: 2px solid red; padding: 15px; border-radius: 10px; margin: 10px; background-color: #f8d7da;'>
                <h3 style='color: red;'>😢 Game Over!</h3>
                <p>The number was {self.secret_number}.</p>
                <p>Your guesses: {self.guesses}</p>
                <p style='color: gray;'>Click 'New Game' to try again!</p>
            </div>
            """
            self.feedback.value = feedback_html
        else:
            # Give hint
            if guess < self.secret_number:
                hint = "📈 Too low! Try higher."
            else:
                hint = "📉 Too high! Try lower."

            self.feedback.value = f"<p>{hint} (Attempt {self.attempts}/{self.max_attempts})</p>"

        self.attempt_label.value = f"Attempts: {self.attempts}/{self.max_attempts}"
        self.guess_input.value = 0

        # Show guess history
        with self.output:
            clear_output()
            if self.guesses:
                guess_history = ", ".join(str(g) for g in self.guesses)
                display(widgets.HTML(f"<p style='font-size: 12px; color: gray;'>Guesses so far: {guess_history}</p>"))

    def show_hint(self, b):
        if self.game_over:
            return

        # Give a better hint
        if self.secret_number <= 20:
            hint = "The number is in the lower range (1-20)."
        elif self.secret_number <= 50:
            hint = "The number is in the middle range (21-50)."
        elif self.secret_number <= 80:
            hint = "The number is in the upper-middle range (51-80)."
        else:
            hint = "The number is in the high range (81-100)."

        if self.secret_number % 2 == 0:
            hint += " It's an even number."
        else:
            hint += " It's an odd number."

        self.feedback.value = f"<p style='color: purple;'>💡 Hint: {hint}</p>"

    def reset_game(self, b):
        self.secret_number = random.randint(1, 100)
        self.attempts = 0
        self.guesses = []
        self.game_over = False
        self.attempt_label.value = f"Attempts: 0/{self.max_attempts}"
        self.feedback.value = "<p style='color: blue;'>🔄 New game started! Guess a number between 1 and 100.</p>"
        self.guess_input.value = 0
        with self.output:
            clear_output()

# Start the game
game = NumberGuessingGame()
```

```python

import random

lower = 1
upper = 100
secret = random.randint(lower, upper) #function to pick random number between range stated
attempts = 0
hint = ""


print("\nNumber Guessing Game!")
print(f"I am thinking of a number between {lower} and {upper}.")

if secret % 2 == 0:
  hint += "It's an even number."
else:
  hint += "It's an odd number."

print("\nHint :", (hint))



while True:
    guess = input(f"\nEnter your guess ({lower}-{upper}): ").strip() #Removes any accidental leading or trailing spaces the user might have typed before or after number

    if not guess.isdigit(): #only return positive whole number
        print("Enter a valid number.")
        attempts -=1
        continue

    guess = int(guess)
    attempts += 1


    if guess < lower or guess > upper:
        print("Your number is out of range")
        attempts -= 1
    elif guess < secret:
        print("Too low! Try again.")
        lower = max(lower, guess + 1)
    elif guess > secret:
        print("Too high! Try again.")
        upper = min(upper, guess - 1)
    else:
        print(f"\nCorrect! The number was {secret}.")
        print(f"You guessed it in {attempts} attempt{'s' if attempts != 1 else ''}!")
        break
```
